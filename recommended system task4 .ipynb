```python
# 🍽️ Restaurant Recommendation System using Content-Based Filtering

```

```python
import pandas as pd
import random
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
users = [f"U{i:03}" for i in range(1, 101)]
restaurants = [f"R{i:03}" for i in range(1, 21)]
cuisines = ["South Indian", "North Indian", "Chinese", "Italian", "Fast Food"]
locations = ["Tirupati", "Puttur", "Chittoor", "Madanapalle"]

data = []
for _ in range(500):
    user = random.choice(users)
    rest = random.choice(restaurants)
    rating = round(random.uniform(1, 5), 1)
    cuisine = random.choice(cuisines)
    location = random.choice(locations)
    data.append([user, rest, rating, cuisine, location])

df = pd.DataFrame(data, columns=["UserID", "RestaurantID", "Rating", "Cuisine", "Location"])
df.to_csv("restaurant_data.csv", index=False)
df.head()
```

```python
df["features"] = df["Cuisine"] + " " + df["Location"]
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df["features"])
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend(restaurant_id):
    try:
        idx = df[df["RestaurantID"] == restaurant_id].index[0]
        scores = list(enumerate(similarity[idx]))
        scores = sorted(scores, key=lambda x: x[1], reverse=True)
        top_indices = [i[0] for i in scores[1:6]]
        return df.iloc[top_indices][["RestaurantID", "Cuisine", "Location", "Rating"]]
    except IndexError:
        return "Restaurant ID not found."
```

```python
recommend("R005")
```
